```python
# RL Circuit (Task 3) – Proper Wiring Diagram
# Signal → Resistor → Inductor → Ground
# Measure Vout across the inductor

import matplotlib.pyplot as plt

fig, ax = plt.subplots()

# Signal input line
ax.plot([0, 2], [0, 0])
ax.text(-0.3, 0, "Signal Input", va='center')

# Resistor (zig-zag)
ax.plot([2, 2.5, 3, 3.5, 4], [0, 0.3, -0.3, 0.3, 0])
ax.text(3, 0.8, "R")

# Wire to inductor
ax.plot([4, 5], [0, 0])

# Inductor (loops)
theta = [i * 3.14 / 20 for i in range(41)]
for i in range(4):
    ax.plot([5 + i*0.4 + 0.2*(1 - __import__("math").cos(t)) for t in theta],
            [0.2*__import__("math").sin(t) for t in theta])

ax.text(6.2, 0.8, "L")

# Wire to ground
ax.plot([6.6, 6.6], [0, -2])
ax.plot([6.1, 7.1], [-2, -2])
ax.plot([6.2, 7.0], [-2.2, -2.2])
ax.plot([6.3, 6.9], [-2.4, -2.4])
ax.text(6.6, -2.7, "Ground", ha='center')

# Vout label across inductor
ax.text(5.5, 0.5, "Vout (Scope)", ha='center')

# Formatting
ax.set_xlim(-1, 8)
ax.set_ylim(-3, 2)
ax.axis('off')

plt.title("RL Transient Circuit (Measure Across Inductor)")
plt.show()
```
